```python
import pandas as pd
import json
from sklearn.svm import SVC
from sklearn.pipeline import make_pipeline
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics import classification_report
```

```python
import json
import pandas as pd


caminho_treino = '/home/cecilia/Documentos/PIBIC/Fase1/dados_svm_bert/train_bio_processado.json'
caminho_val = '/home/cecilia/Documentos/PIBIC/Fase1/dados_svm_bert/val_bio_processado.json'
caminho_teste = '/home/cecilia/Documentos/PIBIC/Fase1/dados_svm_bert/test_bio_processado.json'


with open(caminho_treino, 'r', encoding='utf-8') as f:
    df_train = pd.DataFrame(json.load(f))

with open(caminho_val, 'r', encoding='utf-8') as f:
    df_val = pd.DataFrame(json.load(f))

with open(caminho_teste, 'r', encoding='utf-8') as f:
    df_test = pd.DataFrame(json.load(f))

print("Arquivos carregados com sucesso do Drive!")
```

```python

df_train['target_limpo'] = df_train['aspect_term_completo']
df_val['target_limpo'] = df_val['aspect_term_completo']
df_test['target_limpo'] = df_test['aspect_term_completo']
```

```python

X_train_tokens = [str(token) for sublist in df_train['tokens'] for token in sublist]
y_train_tags = [tag for sublist in df_train['tags'] for tag in sublist]
```

```python


pipeline_svm = make_pipeline(
    TfidfVectorizer(),
    SVC(kernel='linear', C=1.0)
)


pipeline_svm.fit(X_train_tokens, y_train_tags)
```

```python
import pandas as pd
from sklearn.metrics import classification_report


X_test_tokens = [str(token) for sublist in df_test['tokens'] for token in sublist]
y_test_tags = [tag for sublist in df_test['tags'] for tag in sublist]


pipeline_svm.fit(X_train_tokens, y_train_tags)



y_pred_tags = pipeline_svm.predict(X_test_tokens)


labels_bio = ['B-ASP', 'I-ASP', 'O'] 

print("======= MÉTRICAS DE EXTRAÇÃO BIO (FORMATADO) =======") #
print(classification_report(y_test_tags, y_pred_tags, labels=labels_bio, zero_division=0))
```
